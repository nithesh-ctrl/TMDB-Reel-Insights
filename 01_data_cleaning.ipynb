```python
#fill the missing character names
# Verified character names, mapped by cast_row_id
character_mapping = {
    14466: "The Female",
    14467: "The Bad Man",
    14468: "The Dead Woman",
    14469: "Pick-Up Man",
    14470: "First Victim",
    14471: "Leering Man",
    14472: "Second Victim",
    14473: "Man at Club",
    14474: "The Swimmer",
    14475: "Father at Beach"
}

# Fill only missing character names for this movie
mask = (
    (movies_cast["movie_id"] == 97370)
    & movies_cast["character_name"].isna()
)

movies_cast.loc[mask, "character_name"] = (
    movies_cast.loc[mask, "cast_row_id"].map(character_mapping)
)
```
