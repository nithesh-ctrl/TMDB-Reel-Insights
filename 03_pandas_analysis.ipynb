```python
# List all movies with missing or zero budget/revenue values, and decide how these should be treated for the rest of the analysis.

import pandas as pd
import numpy as np

# Load cleaned movie data
movies_df = pd.read_csv("C:\\Users\\acer\\Documents\\TMDB Project\\ReelInsights\\outputs\\cleaned_movies_data.csv")

# Replace zero values with NaN
movies_df["budget"] = movies_df["budget"].replace(0, np.nan)
movies_df["revenue"] = movies_df["revenue"].replace(0, np.nan)

# Calculate median
budget_median = movies_df["budget"].median()
revenue_median = movies_df["revenue"].median()

# Fill missing values with median
movies_df["budget"] = movies_df["budget"].fillna(budget_median)
movies_df["revenue"] = movies_df["revenue"].fillna(revenue_median)


# Check missing values
print("Missing Budget:", movies_df["budget"].isna().sum())
print("Missing Revenue:", movies_df["revenue"].isna().sum())

print(movies_df[["budget", "revenue"]].describe())
```

```python
# List the profit (revenue − budget) and return on investment for every movie that has valid budget and revenue figures.

import pandas as pd

# Load cleaned movie data
movies_df = pd.read_csv("C:\\Users\\acer\\Documents\\TMDB Project\\ReelInsights\\outputs\\cleaned_movies_data.csv")

# Select movies with valid budget and revenue
financial_df = movies_df[
    (movies_df["budget"] > 0) &
    (movies_df["revenue"] > 0)
]

financial_df["profit"] = (
    financial_df["revenue"] - financial_df["budget"]
)

financial_df["roi"] = (
    financial_df["profit"] / financial_df["budget"]
) * 100

financial_df[
    ["movie_id", "title", "budget", "revenue", "profit", "roi"]
].head(10)

print("Movies with valid financial data:", len(financial_df))
```

```python
# List the number of movies released per year, and identify which year had the highest movie output in the dataset.

import pandas as pd

# Step 1: Load the cleaned movies dataset
movies_df = pd.read_csv("C:\\Users\\acer\\Documents\\TMDB Project\\ReelInsights\\outputs\\cleaned_movies_data.csv")

# Step 2: Convert release_date into datetime format
movies_df["release_date"] = pd.to_datetime(
    movies_df["release_date"],
    errors="coerce"
)

# Step 3: Extract the release year
movies_df["release_year"] = movies_df["release_date"].dt.year

# Step 4: Remove records with missing release dates
yearly_movies = movies_df.dropna(
    subset=["release_year"]
).copy()

# Step 5: Convert release_year into integer
yearly_movies["release_year"] = (
    yearly_movies["release_year"].astype(int)
)

# Step 6: Count the number of movies released each year
movies_per_year = (
    yearly_movies.groupby("release_year")["movie_id"]
    .nunique()
    .reset_index(name="total_movies")
)

# Step 7: Sort by release year
movies_per_year = movies_per_year.sort_values(
    by="release_year"
)

# Step 8: Display yearly movie counts
display(movies_per_year)

# Step 9: Identify the year with the highest movie releases
highest_year = movies_per_year.loc[
    movies_per_year["total_movies"].idxmax()
]

print("Year with highest movie releases:",
      highest_year["release_year"])

print("Total movies released:",
      highest_year["total_movies"])
```

```python
# List all movies whose runtime is a statistical outlier compared to the rest of the dataset.

import pandas as pd

# Step 1: Load cleaned movies dataset
movies_df = pd.read_csv("C:\\Users\\acer\\Documents\\TMDB Project\\ReelInsights\\outputs\\cleaned_movies_data.csv")

# Step 2: Select movies with valid runtime
runtime_df = movies_df[
    movies_df["runtime"].notna() &
    (movies_df["runtime"] > 0)
].copy()

# Step 3: Calculate Q1 and Q3
Q1 = runtime_df["runtime"].quantile(0.25)
Q3 = runtime_df["runtime"].quantile(0.75)

# Step 4: Calculate IQR
IQR = Q3 - Q1

# Step 5: Calculate lower and upper limits
lower_limit = Q1 - (1.5 * IQR)
upper_limit = Q3 + (1.5 * IQR)

# Step 6: Identify runtime outliers
runtime_outliers = runtime_df[
    (runtime_df["runtime"] < lower_limit) |
    (runtime_df["runtime"] > upper_limit)
].copy()

# Step 7: Display outlier movies
display(
    runtime_outliers[
        ["movie_id", "title", "runtime"]
    ].sort_values(by="runtime")
)

# Step 8: Print summary
print("First Quartile (Q1):", Q1)
print("Third Quartile (Q3):", Q3)
print("Interquartile Range (IQR):", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Total Runtime Outliers:", len(runtime_outliers))
```

```python
import pandas as pd

# Step 1: Load the datasets
movies_df = pd.read_csv("C:\\Users\\acer\\Documents\\TMDB Project\\ReelInsights\\outputs\\cleaned_movies_data.csv")
movie_genres_df = pd.read_csv("C:\\Users\\acer\\Documents\\TMDB Project\\ReelInsights\\outputs\\cleaned_movie_genres.csv")
genres_df = pd.read_csv("C:\\Users\\acer\\Documents\\TMDB Project\\ReelInsights\\data\\genres.csv")

# Step 2: Merge movie_genres with genres
genre_data = movie_genres_df.merge(
    genres_df,
    on="genre_id",
    how="inner"
)

# Step 3: Merge with movies to get popularity
genre_data = genre_data.merge(
    movies_df[["movie_id", "popularity"]],
    on="movie_id",
    how="inner"
)

# Step 4: Calculate movie count and average popularity by genre
genre_analysis = (
    genre_data.groupby("genre_name")
    .agg(
        total_movies=("movie_id", "nunique"),
        average_popularity=("popularity", "mean")
    )
    .reset_index()
)

# Step 5: Sort by number of movies
genre_analysis = genre_analysis.sort_values(
    by="total_movies",
    ascending=False
)

# Step 6: Display the results
display(genre_analysis)

# Step 7: Identify the most common genre
most_common_genre = genre_analysis.iloc[0]

print("Most Common Genre:", most_common_genre["genre_name"])
print("Number of Movies:", most_common_genre["total_movies"])
print("Average Popularity:", round(most_common_genre["average_popularity"], 2))
```

```python
# List the top 10 most prolific actors by number of movies, along with their average movie rating across those movies.

import pandas as pd

# Step 1: Load datasets
movies_df = pd.read_csv("C:\\Users\\acer\\Documents\\TMDB Project\\ReelInsights\\outputs\\cleaned_movies_data.csv")
cast_df = pd.read_csv("C:\\Users\\acer\\Documents\\TMDB Project\\ReelInsights\\outputs\\cleaned_cast.csv")


# Step 2: Merge cast data with movie ratings
actor_data = cast_df.merge(
    movies_df[["movie_id", "vote_average"]],
    on="movie_id",
    how="inner"
)

# Step 3: Group data by actor
actor_analysis = (
    actor_data.groupby(["person_id", "actor_name"])
    .agg(
        total_movies=("movie_id", "nunique"),
        average_rating=("vote_average", "mean")
    )
    .reset_index()
)

# Step 4: Sort by movie appearances
actor_analysis = actor_analysis.sort_values(
    by="total_movies",
    ascending=False
)

# Step 5: Get top 10 actors
top_10_actors = actor_analysis.head(10).copy()

# Step 6: Round average ratings
top_10_actors["average_rating"] = (
    top_10_actors["average_rating"].round(2)
)

# Step 7: Display results
display(
    top_10_actors[
        ["actor_name", "total_movies", "average_rating"]
    ]
)
```

```python
# List any movie titles that appear more than once in the dataset, along with their release years, to check for duplicates or remakes.

import pandas as pd

# Step 1: Load the cleaned movies dataset
movies_df = pd.read_csv("C:\\Users\\acer\\Documents\\TMDB Project\\ReelInsights\\outputs\\cleaned_movies_data.csv")

# Step 2: Convert release_date to datetime
movies_df["release_date"] = pd.to_datetime(
    movies_df["release_date"],
    errors="coerce"
)

# Step 3: Extract release year
movies_df["release_year"] = movies_df["release_date"].dt.year

# Step 4: Find repeated movie titles
repeated_titles = movies_df[
    movies_df.duplicated(
        subset=["title"],
        keep=False
    )
].copy()

# Step 5: Sort by title and release year
repeated_titles = repeated_titles.sort_values(
    by=["title", "release_year"]
)

# Step 6: Display results
display(
    repeated_titles[
        ["movie_id", "title", "release_year"]
    ]
)

# Step 7: Summary
print("Total records with repeated titles:", len(repeated_titles))

print(
    "Number of repeated movie titles:",
    repeated_titles["title"].nunique()
)
```

```python
# List movies whose popularity score and vote count don't follow the general pattern between the two — i.e., unusually high popularity with a low vote count, or vice versa.

import pandas as pd

# Step 1: Load cleaned movies dataset
movies_df = pd.read_csv("C:\\Users\\acer\\Documents\\TMDB Project\\ReelInsights\\outputs\\cleaned_movies_data.csv")

# Step 2: Select valid popularity and vote count records
analysis_df = movies_df[
    movies_df["popularity"].notna() &
    movies_df["vote_count"].notna() &
    (movies_df["popularity"] >= 0) &
    (movies_df["vote_count"] >= 0)
].copy()

# Step 3: Calculate popularity IQR
popularity_q1 = analysis_df["popularity"].quantile(0.25)
popularity_q3 = analysis_df["popularity"].quantile(0.75)

popularity_iqr = popularity_q3 - popularity_q1

popularity_lower = popularity_q1 - (1.5 * popularity_iqr)
popularity_upper = popularity_q3 + (1.5 * popularity_iqr)

# Step 4: Calculate vote count IQR
vote_q1 = analysis_df["vote_count"].quantile(0.25)
vote_q3 = analysis_df["vote_count"].quantile(0.75)

vote_iqr = vote_q3 - vote_q1

vote_lower = vote_q1 - (1.5 * vote_iqr)
vote_upper = vote_q3 + (1.5 * vote_iqr)

# Step 5: Identify popularity outliers
analysis_df["popularity_outlier"] = (
    (analysis_df["popularity"] < popularity_lower) |
    (analysis_df["popularity"] > popularity_upper)
)

# Step 6: Identify vote count outliers
analysis_df["vote_count_outlier"] = (
    (analysis_df["vote_count"] < vote_lower) |
    (analysis_df["vote_count"] > vote_upper)
)

# Step 7: Identify movies with either type of outlier
outliers = analysis_df[
    analysis_df["popularity_outlier"] |
    analysis_df["vote_count_outlier"]
].copy()

# Step 8: Display outlier movies
display(
    outliers[
        [
            "movie_id",
            "title",
            "popularity",
            "vote_count",
            "popularity_outlier",
            "vote_count_outlier"
        ]
    ].sort_values(
        by="popularity",
        ascending=False
    )
)

# Step 9: Display summary
print("Total movies analysed:", len(analysis_df))

print("Popularity lower limit:", popularity_lower)
print("Popularity upper limit:", popularity_upper)

print("Vote count lower limit:", vote_lower)
print("Vote count upper limit:", vote_upper)

print("Popularity outliers:",
      analysis_df["popularity_outlier"].sum())

print("Vote count outliers:",
      analysis_df["vote_count_outlier"].sum())

print("Total movies with either outlier:",
      len(outliers))
```
